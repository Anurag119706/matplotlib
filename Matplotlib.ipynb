```python
# --------------- 1. LINE PLOT using plt.plot() ---------------

import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(X, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# --------------- 2. SCATTER PLOT using plt.scatter() ---------------

x = [5, 7, 8, 10, 12]
y = [10, 14, 15, 18, 20]

plt.scatter(x, y, color='red', marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
# --------------- 3. BAR CHART using plt.bar() ---------------

# --------------- Vertical bar chart ---------------
x = ['Apple', 'Banana', 'Cherry']
y = [30, 45, 10]

plt.bar(x, y, color='blue')
plt.title("Fruit Sales")
plt.xlabel("Fruit")
plt.ylabel("Quantity Sold")
plt.show()
```

```python
# --------------- 4. HORIZONTAL BAR CHART using plt.barh () ---------------

x = ['Python', 'Java', 'C++']
y = [90, 75, 60]

plt.barh(x, y, color='skyblue')
plt.title("Programming Language Popularity")
plt.xlabel("Score")
plt.ylabel("Language")
plt.show()
```

```python
# --------------- 5. HISTOGRAM using plt.hist() ---------------

import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=20, color='orange', edgecolor='white')
plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
# --------------- 6. PIE CHART using plt.pie() ---------------

sizes = [40, 35, 25]
labels = ['Python', 'Java', 'C++']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Language Popularity Share")
plt.show()
```

```python
# --------------- 7. BOXPLOT using plt.box() ---------------

data = [5, 8, 10, 15, 18, 20, 22, 25, 30]
plt.boxplot(data)
plt.title("Box Plot Example")
plt.ylabel("Values")
plt.show()
```

```python
# --------------- 8. SUBPLOT using plt.subplot() ---------------

x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2)
plt.title("Plot 2")

plt.show()
```

```python
# --------------- 9. LEGEND using plt.legend() ---------------

x = [1, 2, 3, 4]
plt.plot(x, [1, 4, 9, 16], label='Square')
plt.plot(x, [1, 8, 27, 64], label='Cube')

plt.legend()
plt.title("Legend Example")
plt.show()
```

```python
# --------------- 10. GRID LINES using plt.grid() ---------------

x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

plt.plot(x, y)
plt.grid(True)
plt.title("Line Plot with Grid")
plt.show()
```

```python
# --------------- 11. CANVAS FIGURE using plt.figure() ---------------

plt.figure(figsize=(6,4))
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Custom Figure Size")
plt.show()
```

```python
# --------------- 12. SAVEFIGURE using plt.savefig() ---------------

plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Save Plot Example")
plt.savefig("plot_output.png")
plt.show()
```

```python
# --------------- 13. ERRORBAR using plt.errorbar() ---------------

x = [1, 2, 3, 4]
y = [2, 4, 6, 8]
error = [0.2, 0.4, 0.3, 0.5]

plt.errorbar(x, y, yerr=error, fmt='o', color='red', ecolor='yellow', capsize=5)
plt.title("Error Bar Example")
plt.show()
```

```python
# --------------- 14. FILL THE AREA using plt.fill_between () ---------------

import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y, color='skyblue', alpha=0.4)
plt.title("Fill Between Example")
plt.show()
```

```python
# --------------- 15. ANNOTATION using plt.annotate () ---------------

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, color='yellow')
plt.annotate("Peak Point", xy=(5, 10), xytext=(3, 9),
             arrowprops=dict(facecolor='red', edgecolor='red', shrink=0.05))
plt.show()
```
